Methods that are widely used for cleaning the data

```python
data_with_missing = {
    'cars': ["BMW", "Volvo", None, "Tesla", "Ferrari", None,"Ford"],
    'counts': [3, 7, None, 8,1, 9, None]
}

df_missing = pd.DataFrame(data_with_missing)

print(df_missing)
```

```python
#prints out the numbers that are null or missing

df_missing.isnull().sum()
```

```python
#dropna -> method that ignores the row which's got missing valued rows and prints everything except that row

df_missing.dropna(how="all")
```

```python
# deletes the column where even value is missing
df_missing.dropna(axis=1)
```

```python
# fillna() -> fills the values where they are missed
df_missing.fillna("unknown")
```
