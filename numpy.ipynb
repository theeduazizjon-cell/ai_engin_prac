```python
import numpy as np

arr = np.array([1, 3, 5, 7])
print("numpy_array", arr)
```

```python
arr2 = np.array([[1, 3, 5, 7], [4, 6, 8, 9]])
print("numpy_array", arr2)

arr2.shape
```

```python
#dtype -> method printing the type of an array
a = np.array([1, 2, 3])
print(a.dtype)
```

```python
#reshape -> method that changes the shape of the array

a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2,3)

print("reshaped_array:", b)
```

```python
#astype() -> method that changes data type

a = np.array([1,2,3])
b = a.astype(float)

print("float:", b)
```

```python
#np.zeros(), np.ones() -> dummy data

z = np.zeros([2, 3])
o = np.ones([2, 3])

print("zeros:", z)
print("ones:", o)
```

```python
#ndim

a = np.array([[1, 2], [2, 3]])
print("ndim:", a.ndim)
```

```python
#arrange() -> arranging array from the start till the beginning of the numbers thats entered

a = np.arrange(1, 6)
print("arranged:", a.arrange)
```

```python
#sum() , #mean(), #max(), #min()

a = np.array([1,2,3,4])

print("sum:", a.sum())

print("min:", a.min())

print("max:", a.max())

print("mean:", a.mean())
```

```python
#flatten() -> makes 2 or 3d array into one

a = np.array([[1,2], [3,4]])

print("flatten:", a.flatten())
```

```python
#T -> transpose

a = np.array([[1,2], [3,4]])

print("transponse:", a.T)
```

```python
#dot() multiplication of matraxes

a = np.array([[1,2], [3, 4]])
b = np.array([[2, 0], [1, 2]])

print("dot/multiplication:", a.dot(b))
```

```python
#random.rand() randomizing the matrix in the given range

a =np.random.rand(2, 3)

print("random rand:", a)
```
